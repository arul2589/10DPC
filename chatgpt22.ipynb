1: One Class with Multiple Methods

```python
class Application:
    def __init__(self,name,port,status="Stopped"):
        self.name = name
        self.port = port
        self.status = status

    def start(self):
        self.status = "Running"

    def stop(self):
        self.status = "Stopped"

    def get_info(self):
        return (f"{self.name} - {self.port} - {self.status}")

app01 = Application("Tomcat",8080)

print (app01.get_info())

app01.start()

print (app01.get_info())

app01.stop()

print (app01.get_info())
```

2: One Object Using Another Object

```python
class EC2:
    def __init__(self,name,instance_type):
        self.name = name
        self.instance_type = instance_type

class Application:
    def __init__(self,name,port,server):
        self.name=name
        self.port=port
        self.server=server


server1 = EC2("web01","t3.micro")
app1 = Application("Tomcat",8080,server1)


print (f"Application: {app1.name}")
print (f"Port: {app1.port}")
print (f"Server: {app1.server.name}")
print (f"Instance Type: {app1.server.instance_type}")
```

4: Method with Validation

```python
class Application:
    def __init__(self,name,port):
        self.name = name
        self.port = port


    def change_port(self,new_port):
        allowed_ports = [80, 443, 8080]
        if new_port in allowed_ports:
            self.port = new_port
            return "Port updated"

        else:
            return "Invalid port"


app1 = Application("Tomcat",8080)

print (app1.change_port(443))
print (app1.port)

print (app1.change_port(5000))
print (app1.port)
```

5: One Class Creating/Managing Multiple Objects

```python
class EC2:
    def __init__(self,name,instance_type):
        self.name = name
        self.instance_type = instance_type

class Infrastructure:
    def __init__(self):
        self.servers = []

    def add_server(self, server):
        self.servers.append(server)

server1 = EC2("web01", "t3.micro")
server2 = EC2("app01", "t3.medium")
server3 = EC2("db01", "t3.small")

infra = Infrastructure()

infra.add_server(server1)
infra.add_server(server2)
infra.add_server(server3)

for server in infra.servers:
    print (f"{server.name} - {server.instance_type}")
```

9: Count Objects

```python
class EC2:
    def __init__(self,name,instance_type):
        self.name = name
        self.instance_type = instance_type

class Infrastructure:
    def __init__(self):
        self.servers = []

    def add_server(self, server):
        self.servers.append(server)

    def server_count(self):
        return len(self.servers)

    def remove_server(self, server_name):
        for server in self.servers:
            if server.name == server_name:
                self.servers.remove(server)
                return "Server removed"

        return "Server not found"    



server1 = EC2("web01", "t3.micro")
server2 = EC2("app01", "t3.medium")
server3 = EC2("db01", "t3.small")

infra = Infrastructure()

infra.add_server(server1)
infra.add_server(server2)
infra.add_server(server3)

result = infra.server_count()
print(f"Before removal: {result}")

result = infra.remove_server("app01")
print (result)

result = infra.server_count()
print(f"After removal: {result}")
```

self.servers       → Object attribute
[]                 → List
append()           → Add object
remove()           → Remove object
len()              → Count objects
return             → Give value back

```python
class EC2:
    def __init__(self, name, instance_type):
        self.name = name
        self.instance_type = instance_type

class Infrastructure:
    def __init__(self):
        self.servers = []


    def add_server(self,server):
        self.servers.append(server)


    def find_server(self,server_name):
        for server in self.servers:
            if server.name == server_name:
                return server

    def remove_server(self,server_name):
        for server in self.servers:
            if server.name == server_name:
                self.servers.remove(server)
                return "Server removed"

        return "Server not found"

    def server_count(self):
        return len(self.servers)

infra = Infrastructure()

server1 = EC2("web01","t3.micro")
server2 = EC2("app01","t3.medium")
server3 = EC2("db01","t3.small")

infra.add_server(server1)
infra.add_server(server2)
infra.add_server(server3)

result = infra.server_count()
print (f"Total servers: {result}")


result = infra.find_server("db01")

if result is None:
    print ("Server not found")

else:
    print (f"Found: {result.name} - {result.instance_type}")


result = infra.remove_server("app01")

print (result)

result = infra.server_count()
print (f"Total servers: {result}")
```
