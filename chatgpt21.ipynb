Topic 4: Methods
A function written inside a class is called a method.

```python
class Server:
     
     def __init__(self,name,os,port):
          self.name = name
          self.os = os
          self.port = port

     def show_info(self):
         print (f"{self.name} runs {self.os} on port {self.port}")


server1 = Server("web01","Linux",80)


server1.show_info()
          
```

```python
class Server:
    def __init__(self,name,os,port):
        self.name = name
        self.os = os
        self.port = port
    def start(self):
        print(f"Starting server: {self.name}")


server1 = Server("web01","Linux",80)
server2 = Server("app01","Ubuntu",8080)

server1.start()
server2.start()
```

```python
class Server:
    def __init__(self,name,os,port):
        self.name = name
        self.os = os
        self.port = port

    def old_port(self):
        print (f"Old port {self.port}")

    def change_port (self,new_port):
        self.port = new_port
        print (f"New port: {self.port}")
    

server1 = Server("web01","Linux",80)


server1.old_port()
server1.change_port(8080)

```

6: Method with return

```python
class Server:
    def __init__(self,name,os,port):
        self.name = name
        self.os = os
        self.port = port

    def get_info(self):
        return (f"{self.name} - {self.os} - {self.port}")


server1 = Server("web01","Linux",80)

result = server1.get_info()

print (result)
```

8 — Default Values in a Class

we can give status a default value:
Eg: def __init__(self,name,os,status = "Stopped"):

```python
class Server:
    def __init__(self,name,os,status = "Stopped"):
        self.name = name
        self.os = os
        self.status = status


server1 = Server("web01","Linux")
server2 = Server("app01","Ubuntu","Running")

print (f"{server1.name}: {server1.status}")

print (f"{server2.name}: {server2.status}")
        
```

9: Objects in a List

```python
class Server:
    def __init__(self, name, os, status="Stopped"):
        self.name = name
        self.os = os
        self.status = status


server1 = Server("web01","Linux","Running")
server2 = Server("app01","Ubuntu")
server3 = Server("db01","Linux","Running")

servers = [server1,server2,server3]

for server in servers:
    print (f"{server.name} - {server.os} - {server.status}")
```

10: Practical Class

```python
class EC2:
    def __init__(self,name,instance_type,region,status="Stopped"):
        self.name = name
        self.instance_type = instance_type
        self.region = region
        self.status = status

    def start(self):
        self.status = "Running"

    def get_info(self):
        return (f"{self.name} - {self.instance_type} - {self.region} - {self.status}")


ec2 = EC2("web01","t3.micro","ap-south-1")

print (ec2.status)

ec2.start()

result = ec2.get_info()

print (result)


    
```
