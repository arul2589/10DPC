JSON

```python
import json
aws_server = {
    "name" : "web01",
    "region" : "ap-south-1",
    "running": True,
    "status" : None
}

result = json.dumps(aws_server)
print (result)
print (type(result))
```

dumps() → Python to JSON
loads() → JSON to Python

```python
import json

json_data = '{"server":"tomcat01","port":8080,"running":true}'

result = json.loads(json_data)
print(result)
print(type(result))
```

```python
import json

aws_config = {
    "server" : "tomcat01",
    "region" : "ap-south-1",
    "instances" : 2,
    "running" : True
}

with open("aws_config.json","w") as file:
    json.dump(aws_config, file )

print("JSON file created successfully")
```

```python
import json

with open("aws_config.json","r") as file:
    data = json.load(file)

print(f"Server: {data["server"]}")
print(f"Region: {data["region"]}")
print(f"Instances: {data["instances"]}")
print(f"Running: {data["running"]}")
```

```python
import json

with open("aws_config.json","r") as file:
    data = json.load(file)

data["instances"] = 3
data["application"] = "Tomcat"

with open ("aws_config.json","w") as file:
    json.dump(data,file)
print ("JSON configuration updated")
```

Nested JSON

```python
aws = {
    "server" : "web01",
    "network" : {
        "vpc" : "vpc-prod",
        "subnet" : "subnet-web",
        "port" : 8080
    }
}

print (f"Server: {aws["server"]}")
print (f"VPC: {aws["network"]["vpc"]}")
print (f"Subnet: {aws["network"]["subnet"]}")
print (f"Port: {aws["network"]["port"]}")
```

```python
deployment = {
    "server" : "app01",
    "regions" : ["ap-south-1","us-east-1","eu-west-1"]
}

print (f"Server: {deployment["server"]}")
print (f"Primary region: {deployment["regions"][0]}")
print ()
print ("Available regions:")
for region in deployment["regions"]:
    print (region)
```

```python
request = {
    "server": {
        "name": "tomcat01",
        "instances": 2
    },
    "aws": {
        "region": "ap-south-1",
        "instance_type": "t3.micro"
    },
    "application": {
        "name": "Tomcat",
        "port": 8080
    }
}


print (f"Server: {request["server"]["name"]}")
print (f"Instances: {request["server"]["instances"]}")
print (f"Region: {request["aws"]["region"]}")
print (f"Instance type: {request["aws"]["instance_type"]}")
print (f"Application: {request["application"]["name"]}")
print (f"Port: {request["application"]["port"]}")

cost = request["server"]["instances"]*500
print (f"Estimated cost: {cost}")
```

```python
import json

deployment = {
    "server": "app01",
    "instances": 3,
    "region": "ap-south-1",
    "production": True
}

with open("deployment.json","w") as file:
    json.dump(deployment,file,indent=4)

with open("deployment.json","r") as file:
    data = json.load(file)

print (f"Server: {data["server"]}")
print (f"Instances: {data["instances"]}")
print (f"Region: {data["region"]}")
print (f"Production: {data["production"]}")
```

```python
import json

request_json = '''{
    "server": "web01",
    "instances": 2,
    "region": "ap-south-1",
    "application": {
        "name": "Tomcat",
        "port": 8080
    }
}'''

json_data = json.loads(request_json)

print (f"Server: {json_data["server"]}")
print (f"Instances: {json_data["instances"]}")
print (f"Region: {json_data["region"]}")
print (f"Application: {json_data["application"]["name"]}")
print (f"Port: {json_data["application"]["port"]}")
print ()

cost = json_data["instances"]*750
print (f"Estimated cost: {cost}")
print ()

if json_data["application"]["port"] == 8080:
    print ("Tomcat port detected")

else:
    print ("Different application port")
```
